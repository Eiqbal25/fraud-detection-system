# EDA: Auto Insurance Claims Fraud Dataset

Exploratory analysis only. The actual cleaning/feature-engineering/modeling pipeline lives in `src/`, not here -- this notebook is scratch space for understanding the data, not the source of truth.

```python
import sys
sys.path.insert(0, '..')

import pandas as pd
import matplotlib.pyplot as plt

from src.config import RAW_DATA_PATH

df = pd.read_csv(RAW_DATA_PATH, na_values=['?'])
df.shape
```

```python
df['fraud_reported'].value_counts(normalize=True)
```

Fraud is the minority class at ~24.7% -- imbalanced, though not extremely so. This shapes the modeling choices in `src/models/train.py` (class weighting + SMOTE, both implemented and compared).

```python
df.isna().sum()[df.isna().sum() > 0]
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df.boxplot(column='total_claim_amount', by='fraud_reported', ax=axes[0])
df.boxplot(column='age', by='fraud_reported', ax=axes[1])
plt.suptitle('')
plt.tight_layout()
plt.show()
```

```python
pd.crosstab(df['incident_severity'], df['fraud_reported'], normalize='index')
```

`incident_severity` == 'Major Damage' has a visibly higher fraud rate than other severities -- this shows up later as the single strongest SHAP feature for the final model (see `results/shap_plots/global_importance.png`).

One quirk worth flagging: `insured_hobbies` also turns out to be a strong SHAP driver (particular hobbies like chess/cross-fit correlate with fraud in this dataset). That's a well-known artifact of this specific toy Kaggle dataset rather than a generalizable real-world signal -- documented honestly in the main README rather than presented as a genuine fraud indicator.
